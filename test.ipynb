```python
import torch

max_seq_len = 256

positions = torch.arange(max_seq_len)
print(positions)
print(positions.shape)
```

```python
head_dim = 16

pair_indices = torch.arange(head_dim // 2)
print(pair_indices)
```

```python
rope_base = 10000

thetas = rope_base ** (-2 * pair_indices / head_dim)
print(thetas)
```

```python
print(positions.unsqueeze(0).shape)
print(thetas.unsqueeze(1).shape)
angles = positions.unsqueeze(0) * thetas.unsqueeze(1)
angles.shape
```
